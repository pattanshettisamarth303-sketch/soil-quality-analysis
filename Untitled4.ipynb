```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

# 1. Image preprocessing and augmentation
IMG_SIZE = (224, 224)
BATCH_SIZE = 16

train_datagen = ImageDataGenerator(
    rescale=1.0/255.0,
    rotation_range=20,
    horizontal_flip=True,
    vertical_flip=True,
    zoom_range=0.2,
    shear_range=0.1
)

val_test_datagen = ImageDataGenerator(rescale=1.0/255.0)

train_generator = train_datagen.flow_from_directory(
    'data/processed/images/train',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)

val_generator = val_test_datagen.flow_from_directory(
    'data/processed/images/val',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# 2. Build Transfer Learning Model with Pretrained ResNet50
base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False  # Freeze pretrained layers

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(128, activation='relu')(x)
x = Dropout(0.3)(x)
predictions = Dense(train_generator.num_classes, activation='softmax')(x)

model = Model(inputs=base_model.input, outputs=predictions)

# 3. Configure optimizer and loss
model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print(f"Model compiled with {train_generator.num_classes} output classes.")
```

```python
import matplotlib.pyplot as plt
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# 1. Setup Callbacks to prevent overfitting and save the best weights
callbacks = [
    EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True),
    ModelCheckpoint('best_soil_resnet50.keras', monitor='val_accuracy', save_best_only=True)
]

# 2. Train the CNN model
EPOCHS = 20
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=EPOCHS,
    callbacks=callbacks
)

# 3. Plot Training & Validation Accuracy and Loss Curves
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy', marker='o')
plt.plot(history.history['val_accuracy'], label='Val Accuracy', marker='o')
plt.title('Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss', marker='o')
plt.plot(history.history['val_loss'], label='Val Loss', marker='o')
plt.title('Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix

# 1. Load Test Data Generator
test_generator = val_test_datagen.flow_from_directory(
    'data/processed/images/test',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# 2. Generate Predictions
y_probs = model.predict(test_generator)
y_pred = np.argmax(y_probs, axis=1)
y_true = test_generator.classes
class_names = list(test_generator.class_indices.keys())

# 3. Calculate Accuracy, Precision, Recall, and F1-score
print("\n" + "="*50)
print(" Week 3 CNN Evaluation Report (ResNet-50)")
print("="*50)
print(classification_report(y_true, y_pred, target_names=class_names))

# 4. Generate & Plot Confusion Matrix
cm = confusion_matrix(y_true, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Soil Classification Confusion Matrix')
plt.ylabel('Actual Soil Class')
plt.xlabel('Predicted Soil Class')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 5. Save the Final Model Artifact
model.save('soil_resnet50_final.keras')
print("\nModel saved successfully as 'soil_resnet50_final.keras'")
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix

IMG_SIZE = (224, 224)
BATCH_SIZE = 8

# 1. Image preprocessing with proper MobileNet normalization [-1, 1]
train_datagen = ImageDataGenerator(
    preprocessing_function=tf.keras.applications.mobilenet_v2.preprocess_input,
    rotation_range=25,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True
)

val_test_datagen = ImageDataGenerator(
    preprocessing_function=tf.keras.applications.mobilenet_v2.preprocess_input
)

train_gen = train_datagen.flow_from_directory(
    'data/processed/images/train',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=True
)

val_gen = val_test_datagen.flow_from_directory(
    'data/processed/images/val',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

test_gen = val_test_datagen.flow_from_directory(
    'data/processed/images/test',
    target_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    shuffle=False
)

# 2. Build MobileNetV2 Architecture
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
base_model.trainable = False

x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(64, activation='relu')(x)
x = Dropout(0.2)(x)
output = Dense(5, activation='softmax')(x)

mob_model = Model(inputs=base_model.input, outputs=output)

mob_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# 3. Train
history = mob_model.fit(
    train_gen,
    validation_data=val_gen,
    epochs=12,
    verbose=1
)

# 4. Evaluate Test Set & Metrics
y_probs = mob_model.predict(test_gen)
y_pred = np.argmax(y_probs, axis=1)
y_true = test_gen.classes
class_names = list(test_gen.class_indices.keys())

print("\n" + "="*50)
print(" CNN Final Evaluation Report")
print("="*50)
print(classification_report(y_true, y_pred, target_names=class_names, zero_division=0))

# 5. Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.title('Soil Classification Confusion Matrix')
plt.ylabel('Actual Soil Class')
plt.xlabel('Predicted Soil Class')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 6. Save Model Deliverable
mob_model.save('soil_vision_best_model.keras')
print("\nFinal trained model saved as 'soil_vision_best_model.keras'")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import shap
import tensorflow as tf

# ==========================================
# 1. Grad-CAM for CNN Vision Model
# ==========================================
def make_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):
    grad_model = tf.keras.models.Model(
        inputs=[model.inputs],
        outputs=[model.get_layer(last_conv_layer_name).output, model.output]
    )
    with tf.GradientTape() as tape:
        last_conv_layer_output, preds = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]

    grads = tape.gradient(class_channel, last_conv_layer_output)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()

# Find last conv layer in MobileNetV2 base
last_conv_layer_name = 'Conv_1'

# Grab one sample test image
sample_img_batch, sample_label_batch = next(iter(test_gen))
sample_img = sample_img_batch[0:1]

heatmap = make_gradcam_heatmap(sample_img, mob_model, last_conv_layer_name)

# Rescale heatmap and superimpose
heatmap_resized = cv2.resize(heatmap, (224, 224))
heatmap_colored = np.uint8(255 * heatmap_resized)
heatmap_colored = cv2.applyColorMap(heatmap_colored, cv2.COLORMAP_JET)

# Display Grad-CAM
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
# De-normalize image for visualization
display_img = (sample_img[0] + 1) / 2.0
plt.imshow(np.clip(display_img, 0, 1))
plt.title("Original Soil Sample")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(np.clip(display_img, 0, 1))
plt.imshow(heatmap_resized, cmap='jet', alpha=0.5)
plt.title("Grad-CAM Attention Heatmap")
plt.axis('off')
plt.tight_layout()
plt.show()

# ==========================================
# 2. SHAP Explanations for Structured Model
# ==========================================
import shap
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier

# 1. Fit Random Forest for multi-class SHAP TreeExplainer compatibility
rf_model = RandomForestClassifier(n_estimators=100, max_depth=6, random_state=42)
rf_model.fit(X_train, y_train)

# 2. Compute SHAP values
sample_X = X_test.iloc[:100]
explainer = shap.TreeExplainer(rf_model)
shap_values = explainer(sample_X)

# 3. Plot Feature Importance Summary across all soil/climate parameters
plt.figure(figsize=(10, 6))
# For multi-class, shap_values has shape (samples, features, classes)
shap.summary_plot(shap_values[:, :, 0], sample_X, plot_type="bar", show=False)
plt.title("SHAP Feature Importance (Soil & Climate Parameters)")
plt.xlabel("Mean |SHAP Value| (Impact on Model Output)")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np

def hybrid_soil_analysis(image_tensor, soil_tabular_dict, vision_model, tabular_model, class_labels):
    # 1. Vision Prediction & Confidence
    img_probs = vision_model.predict(image_tensor, verbose=0)[0]
    img_class_idx = np.argmax(img_probs)
    img_soil_type = class_labels[img_class_idx]
    img_confidence = img_probs[img_class_idx] * 100

    # 2. Structured Tabular Model Prediction
    tabular_df = pd.DataFrame([soil_tabular_dict])
    tab_probs = tabular_model.predict_proba(tabular_df)[0]
    tab_class_idx = np.argmax(tab_probs)
    recommended_crop = tabular_model.classes_[tab_class_idx]
    tab_confidence = tab_probs[tab_class_idx] * 100

    # 3. Soil Health Index Scoring (0 to 100)
    ph = soil_tabular_dict['ph']
    n = soil_tabular_dict['N']
    p = soil_tabular_dict['P']
    k = soil_tabular_dict['K']

    # Score component calculations
    ph_score = max(0, 100 - abs(6.5 - ph) * 20)
    npk_balance = min(100, ((n / 100) + (p / 50) + (k / 50)) / 3 * 100)
    soil_health_score = round(0.5 * ph_score + 0.5 * npk_balance, 2)

    # 4. Generate Final Hybrid Report
    print("=" * 60)
    print("          HYBRID SOIL ASSESSMENT SYSTEM REPORT")
    print("=" * 60)
    print(f"Detected Soil Type (Vision)    : {img_soil_type} ({img_confidence:.2f}% confidence)")
    print(f"Recommended Crop (Structured)  : {recommended_crop.capitalize()} ({tab_confidence:.2f}% confidence)")
    print(f"Overall Soil Health Score      : {soil_health_score}/100")

    # Nutrient Deficiency Warnings
    warnings = []
    if n < 50: warnings.append("Nitrogen (N) Deficient")
    if p < 30: warnings.append("Phosphorus (P) Deficient")
    if k < 20: warnings.append("Potassium (K) Deficient")
    if ph < 5.5: warnings.append("Soil is acidic")
    elif ph > 7.5: warnings.append("Soil is alkaline")

    if warnings:
        print(f"Alerts & Deficiencies          : {', '.join(warnings)}")
    else:
        print("Alerts & Deficiencies          : Optimal nutrient balance")
    print("=" * 60)

# Run hybrid assessment on sample data
test_soil_sample = {
    'N': 85, 'P': 40, 'K': 45,
    'temperature': 24.5, 'humidity': 80.0,
    'ph': 6.5, 'rainfall': 200.0
}

hybrid_soil_analysis(
    image_tensor=sample_img,
    soil_tabular_dict=test_soil_sample,
    vision_model=mob_model,
    tabular_model=gb_model,
    class_labels=class_names
)
```
